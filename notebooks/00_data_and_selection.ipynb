# 00 — The data and the selection

**Paper element.** The measured detection efficiency and candidate lists behind
Letter Fig. 2; the Letter figure itself is drawn by
`scripts/paper_fig_data_spectrum.py --stem efficiency` from the same products
(`scripts/paper_fig_efficiency.py` draws the supporting three-mode efficiency
comparison, which is not a Letter figure).

**What this notebook is.** What the instrument delivered, before any physics
model touches it: the per-mode detection efficiency $\varepsilon(q)$, the
exposure and duty cycle, every reconstructed blip in the run, the night-selected
candidate lists the limit is actually set on — and *why* the selection is the
night one.

**It runs from a fresh checkout.** Everything below is read from the tracked
data release (`release/luhdm_datarelease_v9_A_f1_atm.h5`, whose `/detector`
group carries the efficiency table, the candidate lists and all the
reconstructed blips) and from the committed efficiency tables in
`luhdm/reference_data/`. One block is the exception: the **full-run exposure and
duty cycle** need the raw instrument file
`data/fit_data_temp_lockin_transients_selected.hdf5`, which is far too large to
keep in git and is not distributed with the code. That block is guarded on the
file's presence — with it, the notebook additionally re-derives the blip
momenta from the raw transients and checks them against the released ones;
without it, the block prints one line and is skipped.

The measured products behind the search, following the recipe in
`data/analyse_transients_data.ipynb` (also not distributed):

1. **Efficiency curves** — the analytic per-segment detection efficiency for
   mode $n$ under the calibrated $\chi^2_3$ hypothesis, averaged over the
   night-selected segments. These curves are **marginalised over the impulse
   arrival phase $w$** — where inside the sampling interval the kick happens to
   land (detector group, August 2026) — and supersede the earlier product, which
   fixed $w = 1$, a single arrival phase. Averaging over the phase costs
   efficiency through the turn-on and leaves the saturated plateau essentially
   unchanged, so the fixed phase was the optimistic one. Section "Which
   efficiency product" plots all the variants against each other.
2. **Blips** — the `upcrossings_{n}` (contiguous runs of transient statistic
   $q>100$). Each blip's kick amplitude `transient_dv_hat` is turned into an
   imparted momentum in eV/$c$; every up-crossing reconstructed above the
   **100 GeV (0.1 TeV) reconstruction threshold** is written to the per-mode
   `data_mode{n}.txt` candidate lists.

**Two thresholds, and they are not the same number.** 100 GeV is the
*reconstruction* threshold: the lowest imparted momentum at which an
up-crossing is reconstructed at all, and therefore what the stored lists
contain — every reconstructed candidate above it, nothing removed. The
*analysis window* the limit is set in opens an order of magnitude higher, at
`config.Q_THRESH` = 1 TeV. That edge is a stated momentum threshold, not a
point read off the efficiency curve: with the arrival-phase-marginalised
curves the window opens where mode 1 is at $\varepsilon = 0.17$ (half
efficiency at 1.21 TeV) and mode 2 at 0.59 (half at 0.95 TeV), while mode 3 is
still on the toe of its turn-on there ($\varepsilon \approx 0$ at 1 TeV, half
at 5.54 TeV). Whatever the efficiency is at a given momentum is folded into
the rate at that momentum, so the window fixes which momenta the search speaks
for, not how well it responds inside them. The limit is set on the subset
of these candidates that falls inside that window: all 8 mode-1 night
candidates (1.52–12.8 TeV) and all 126 mode-3 ones, but 24 of the 26 mode-2
candidates — the two sub-TeV mode-2 impulses, at 554 and 946 GeV, sit below the
window and take no part in setting the limit.

**Why the night selection.** The raw file holds the **full run** — all 3122
segments, 66/99/443 blips over ≈ 470 h. The analysis runs on the **night-time
selection**: segments starting 19:00–07:00 local, minus 19–20 Jan 2026
(elevated lab vibration) and the segments with the lock-in calibration drive on.
That is 8/26/126 candidates over 219.66 h, 47% of the run's live-time. The
daytime hours are the ones the building is occupied in and the ones whose
transient population is visibly vibration-dominated; the two vetoed days and the
drive-on segments are known instrumental states rather than statistical
outliers. Excluding them costs a little over half the exposure and buys a
population the efficiency model was calibrated on — which is the trade the
efficiency curves above are averaged over.

> **Selection caveat.** The night selection is not reconstructable from the raw
> file (the drive-on veto is not a column there), so the candidate lists are
> **read** from `data_mode{n}.txt` — and cross-checked against the release —
> rather than recomputed here. The efficiency curves are read too, from
> `luhdm.efficiency` (`LUHDM_EFFICIENCY_NPZ`), exactly as the detector group
> delivered them. The one thing still shown for the full run is the **blip
> momentum spectrum** below, which needs the sub-window blips; those ship in the
> release's `/detector/all_blips_mode{n}`, so that figure needs no raw file
> either. It is labelled as full-run wherever it appears.

Natural-unit momentum: $p_{[\rm eV/}c] = p_{\rm SI}\,(c/e)$ (SI kg m s$^{-1}$ $\to$ eV/$c$; momentum, not energy).

These three products — the per-mode efficiency $\varepsilon(q)$, the candidate
lists `data_mode{n}.txt`, and the live-time below — are the inputs to every
other notebook here: [`01_the_limit.ipynb`](01_the_limit.ipynb) (the result),
[`02_how_the_limit_is_made.ipynb`](02_how_the_limit_is_made.ipynb) (the
pipeline) and
[`04_composite_benchmark.ipynb`](04_composite_benchmark.ipynb) (the
atmosphere-off benchmark).

```python
%matplotlib inline
from pathlib import Path
import os as _os

import matplotlib.pyplot as plt
import numpy as np

from luhdm import config, efficiency as _eff
from luhdm import release

for _d in ("png", "svg", "pdf"):
    _os.makedirs(_d, exist_ok=True)

# The released cube supplies every measured product this notebook shows: the
# efficiency table, the candidate lists and all the reconstructed blips live in
# its /detector group. That is what lets this notebook run from a fresh
# checkout; before the v9 release it was the one notebook that could not.
rel = release.open_release("../release/luhdm_datarelease_v9_A_f1_atm.h5")


def save_figure(fig, name, **kw):
    # every figure lands in png/, svg/ and pdf/ with a notebook-number prefix,
    # stamped with the data-release version whose /detector group it came from
    fig.text(0.995, 0.004, f"POLONAISE UHDM data release {rel.version_tag}",
             ha="right", va="bottom", fontsize=6, color="#9a9a9a")
    for ext in ("png", "svg", "pdf"):
        fig.savefig(f"{ext}/{name}.{ext}", dpi=140, **kw)


DATA = Path("..") / "data"
RAW = DATA / "fit_data_temp_lockin_transients_selected.hdf5"
HAVE_RAW = RAW.exists()          # the exposure/duty-cycle block needs this
MODES = (1, 2, 3)
COLORS = {1: "#1f77b4", 2: "#ff7f0e", 3: "#2ca02c"}  # one colour per mode
Q_STAT_THRESHOLD = 100   # transient statistic q defining an up-crossing (a blip)
Q_RECO_TEV = 0.1         # reconstruction threshold [TeV/c]: every up-crossing
                         # above it is stored in the candidate lists
Q_WINDOW_TEV = config.Q_THRESH / 1e3   # analysis window opens here [TeV/c];
                                       # the limit uses only candidates above it

# momentum unit conversions
C_LIGHT = 2.99792458e8       # m/s
E_CHARGE = 1.602176634e-19   # J/eV
EV_PER_SI = C_LIGHT / E_CHARGE          # p[eV/c] = p[kg m/s]*this (momentum, NU)
TEV_PER_SI = EV_PER_SI / 1e12           # p[TeV/c] = p[kg m/s]*this

# ONE efficiency product supplies everything below: the committed table
# luhdm/reference_data/efficiency_curves.npz, the same file the limit scans
# load through luhdm.efficiency and the same sha256 the release cubes record.
# It carries the momentum grid in GeV together with the exact SI->GeV factor it
# was converted with, so the detector group's SI momentum bins are recovered
# from it rather than read out of a second, superseded npz.
_TABLE = Path(_eff.table_path())
with np.load(_TABLE) as _z:
    GEV_PER_SI = float(_z["gev_per_si"])                     # p[GeV] / p[kg m/s]
    DV_BINS = {n: _z[f"q_gev_{n}"] / GEV_PER_SI for n in MODES}   # p [kg m/s]

# dv_bins_n = |dv|[volts] * K_n, with |dv| = logspace(-6,-3); recover K_n (SI/volt)
_dv_volts = np.logspace(-6, -3, 400)
K_VOLT_TO_SI = {n: float(np.median(DV_BINS[n] / _dv_volts)) for n in MODES}
print("volt->momentum K_n [kg m s^-1 / V]:",
      {n: f'{K_VOLT_TO_SI[n]:.3e}' for n in MODES})

# K_n is only meaningful if the grid really is that logspace times a constant,
# which is what makes the median a calibration rather than an average. Checked
# on the table itself, so the check needs no second product either.
for n in MODES:
    assert np.allclose(DV_BINS[n], _dv_volts * K_VOLT_TO_SI[n], rtol=1e-12), \
        f"mode {n}: dv_bins is not logspace(-6,-3) x a constant; recheck K_n"
print("dv_bins = logspace(-6, -3, 400) x K_n to 1e-12 -> K_n is exact")

# ---------------------------------------------------------------------------
# The three analysis products, read rather than recomputed, and cross-checked
# against the release so the notebook and the cube cannot describe different
# data:
#   * luhdm.efficiency.load_curve -> LUHDM_EFFICIENCY_NPZ (committed table,
#     the same file the K_n calibration above came from)  vs  /detector/eff_*
#   * data_mode{n}.txt beside this notebook (night candidates) vs /detector/events_*
#   * all reconstructed blips: /detector/all_blips_mode{n}
# ---------------------------------------------------------------------------
EFF_TABLE = {n: _eff.load_curve(n, df=3) for n in MODES}   # (q_GeV, epsilon)
CANDIDATES = {n: np.sort(np.loadtxt(f"data_mode{n}.txt")) for n in MODES}
ALL_MOMENTA = {n: np.sort(rel.all_blips(n)) for n in MODES}     # [eV], full run
for n in MODES:
    q_c, e_c = rel.efficiency_curve(n, df=3)
    assert np.allclose(EFF_TABLE[n][0], q_c) and np.allclose(EFF_TABLE[n][1], e_c), \
        f"mode {n}: the committed efficiency table differs from the release's"
    assert np.allclose(CANDIDATES[n] / 1e9, np.sort(rel.events(n))), \
        f"mode {n}: data_mode{n}.txt differs from the release's candidate list"

_tbl = Path(_eff.table_path())
_repo = Path(_eff.__file__).resolve().parents[1]   # the package sits in the repo
print("efficiency table       :",
      _tbl.relative_to(_repo) if _tbl.is_relative_to(_repo) else _tbl.name)
print("  (matches the release's /detector copy, and its recorded sha256)")
print("night candidates       :",
      {n: CANDIDATES[n].size for n in MODES})
print("  of those, in window  :",
      {n: int((CANDIDATES[n] >= Q_WINDOW_TEV * 1e12).sum()) for n in MODES})
print("all reconstructed blips:", {n: ALL_MOMENTA[n].size for n in MODES},
      " (full run, from /detector/all_blips_mode{n})")
print("raw instrument file    :",
      f"{RAW} -- present" if HAVE_RAW else f"{RAW} -- absent (one block skips)")
```

## Units & conversion: volts → momentum

Every observable here is a **momentum** (the momentum transfer $q$). In natural
units momentum is written in eV but is strictly **eV/$c$** — not energy.

The HDF5 gives each blip's kick amplitude `transient_dv_hat` in lock-in “volts”. The chain to momentum is

$$ |\hat{dv}|\;[\mathrm{V}] \;\xrightarrow{\times K_n}\; p\;[\mathrm{kg\,m\,s^{-1}}] \;\xrightarrow{\times\,c/e}\; p\;[\mathrm{eV}/c] \;\xrightarrow{\div 10^9}\; q\;[\mathrm{GeV}/c]. $$

- **$K_n$** (volts → SI momentum) $= \mathrm{conversion\_m\_V}_n \cdot f_n \cdot 2\pi \cdot m_{\rm sensor}$ (volts→displacement→velocity→momentum).
  We recover it empirically rather than hardcoding those constants: the
  efficiency product's momentum bins are exactly
  `np.logspace(-6, -3, 400) * K_n`, so
  `K_VOLT_TO_SI[n] = median(dv_bins_n / dv_volts)` — values ≈ 1.7e-11,
  2.0e-11, 1.4e-10 kg m s$^{-1}$/V. **Every quantity in this notebook comes from
  one product**, the committed table `luhdm/reference_data/efficiency_curves.npz`
  that `luhdm.efficiency` loads and the release cubes record the sha256 of: the
  curves plotted below are its `eff_{n}_df3`, and the bins the calibration uses
  are its `q_gev_{n}` divided by the `gev_per_si` factor stored beside them. The
  momentum grid has not moved across efficiency products, so the blip momenta
  below are calibrated exactly as they were.
- **$c/e = 1.8711\times10^{27}$** (`EV_PER_SI`) turns SI momentum into eV/$c$ —
  the single factor of $c$ ($1\,\mathrm{GeV}/c = 5.344\times10^{-19}$ kg m/s).
- The per-mode `data_mode{n}.txt` store $p$ in eV/$c$; the limit scan reads them,
  divides by $10^9$ to get $q$ in GeV/$c$, and works inside its analysis window
  from `config.Q_THRESH` = 1 TeV upwards.

## Efficiency curves

Drawn from the table the limit scans themselves load (`luhdm.efficiency`), so
the curve shown here and the curve folded into the rate are the same object.
These are the night-selection curves **marginalised over the impulse arrival
phase $w$** (detector group, August 2026): they replace the earlier product,
which fixed $w = 1$, and the change is confined to the turn-on — mode 1 reaches
half efficiency at 1.21 TeV rather than 0.98 TeV, and $\varepsilon$ at 1 TeV
drops from 0.55 to 0.17. The dashed line marks where the analysis window opens, so
the turn-on and the window can be read against each other.

```python
fig, ax = plt.subplots(figsize=(7.5, 5))
for n in MODES:
    q_gev, eff3 = EFF_TABLE[n]              # the table the limit scans load
    ax.plot(q_gev / 1e3, eff3, color=COLORS[n], lw=2.0, label=f"mode {n}")
ax.set_xscale("log")
ax.set_xlim(0.15, 20)
ax.set_ylim(-0.02, 1.03)
ax.axhline(0.5, color="0.7", lw=0.8, ls=":", zorder=0)
ax.axvline(Q_WINDOW_TEV, color="0.4", lw=1.0, ls="--", zorder=0)
ax.text(Q_WINDOW_TEV * 0.95, 1.0, f"analysis window\nopens at {Q_WINDOW_TEV:g} TeV",
        ha="right", va="top", fontsize=8, color="0.3")
ax.set_xlabel(r"Imparted momentum  [TeV/$c$]")
ax.set_ylabel("Detection efficiency")
ax.grid(True, which="both", alpha=0.15)
ax.legend(loc="center left", frameon=False, fontsize=9)
sec = ax.secondary_xaxis("top",
                         functions=(lambda t: t / TEV_PER_SI,
                                    lambda s: s * TEV_PER_SI))
sec.set_xlabel(r"Imparted momentum, $p$  [m kg s$^{-1}$]")
fig.tight_layout()
save_figure(fig, "00_efficiency_curves")
plt.show()

# 50% points and the efficiency where the window opens, for the record
for n in MODES:
    q_gev, eff3 = EFF_TABLE[n]
    q50 = float(np.interp(0.5, eff3, q_gev))
    eps_edge = float(np.interp(config.Q_THRESH, q_gev, eff3))
    print(f"mode {n}: eps = 0.5 at q = {q50:7.1f} GeV = {q50 / 1e3:.3f} TeV; "
          f"eps({config.Q_THRESH / 1e3:g} TeV) = {eps_edge:.3f}; "
          f"plateau = {eff3.max():.3f}")
```

### Which efficiency product

Three products exist for the same measurement, and the release uses exactly one
of them. The differences are confined to the turn-on, which is where the
analysis window sits, so it is worth seeing them against each other:

- **night, marginalised $w$** — `efficiency_curves_marginalise_w_night.npz`,
  copied to `luhdm/reference_data/efficiency_curves.npz` and recorded by sha256
  in every v9 cube. Averaged over the night-selected segments *and* over the
  impulse arrival phase. **This is the release's curve**, and the one the figure
  above draws.
- **full run, marginalised $w$** — the same phase average over every segment in
  the run. It is the table the v8 cubes were built against, kept committed so
  the v8 → v9 change is inspectable. Averaging in the daytime segments shifts
  mode 1's half-efficiency point up by about 15 GeV.
- **fixed $w = 1$** — the earlier product, a single arrival phase. It is the
  optimistic one: it puts mode 1's half-efficiency point at 0.98 TeV and
  $\varepsilon$(1 TeV) at 0.55, against 1.21 TeV and 0.17 for the release curve.
  That factor of three at the window edge is why the v9 limits are not simply
  the v8 ones re-labelled. This product ships in `data/`, so it is drawn only
  when `data/` is present.

The ratio panel makes the point in one line: away from the turn-on all three
products agree to well under a percent, and everything that separates them is
inside the decade around the window edge.

```python
REF_DIR = Path(_eff.__file__).resolve().parent / "reference_data"
VARIANTS = [
    ("night, marginalised $w$ (the release)", REF_DIR /
     "efficiency_curves_marginalise_w_night.npz", "#1f77b4", "-"),
    ("full run, marginalised $w$ (v8)", REF_DIR /
     "efficiency_curves_marginalise_w.npz", "#8c564b", (0, (5, 3))),
    ("fixed $w$ = 1 (superseded)", DATA / "selected_data_efficiency_curves.npz",
     "#7f7f7f", (0, (1, 1))),
]


def variant_curve(path, mode, df=3):
    """(q_GeV, epsilon) from a detector-group npz, or None if it is absent."""
    if not path.exists():
        return None
    with np.load(path) as z:
        return z[f"dv_bins_{mode}"] * GEV_PER_SI, z[f"efficiency_{mode}_{df}"]


MODE_SHOW = 1
fig, (a0, a1) = plt.subplots(2, 1, figsize=(7.5, 6.2), sharex=True,
                             gridspec_kw=dict(height_ratios=[2.2, 1.0],
                                              hspace=0.08))
q_ref, e_ref = EFF_TABLE[MODE_SHOW]
print(f"mode {MODE_SHOW} efficiency products:")
for lab, path, col, ls in VARIANTS:
    cur = variant_curve(path, MODE_SHOW)
    if cur is None:
        print(f"  {lab:38s}: not on this machine ({path.name})")
        continue
    q, e = cur
    a0.plot(q / 1e3, e, color=col, lw=2.0, ls=ls, label=lab)
    a1.plot(q / 1e3, np.divide(e, e_ref, out=np.full_like(e, np.nan),
                               where=e_ref > 1e-6),
            color=col, lw=1.8, ls=ls)
    print(f"  {lab:38s}: 50% at {float(np.interp(0.5, e, q)):7.1f} GeV, "
          f"eps(1 TeV) = {float(np.interp(config.Q_THRESH, q, e)):.3f}, "
          f"plateau {e.max():.3f}")

# the release's own curve is the reference of the ratio panel
assert np.allclose(variant_curve(VARIANTS[0][1], MODE_SHOW)[1], e_ref), \
    "the committed table is no longer the night, marginalised-w product"

a0.set_xscale("log")
a0.set_xlim(0.3, 20)
a0.set_ylim(-0.02, 1.05)
a0.axhline(0.5, color="0.7", lw=0.8, ls=":", zorder=0)
a0.axvline(Q_WINDOW_TEV, color="0.4", lw=1.0, ls="--", zorder=0)
a0.text(Q_WINDOW_TEV * 0.95, 1.02, f"analysis window\nopens at "
        f"{Q_WINDOW_TEV:g} TeV", ha="right", va="top", fontsize=8, color="0.3")
a0.set_ylabel("Detection efficiency")
a0.set_title(f"mode {MODE_SHOW}: the three efficiency products", fontsize=10)
a0.grid(True, which="both", alpha=0.15)
a0.legend(loc="center left", frameon=False, fontsize=8.5)

a1.axhline(1.0, color="#1f77b4", lw=1.4)
a1.axvline(Q_WINDOW_TEV, color="0.4", lw=1.0, ls="--", zorder=0)
a1.set_ylim(0, 4.2)
a1.set_xlabel(r"Imparted momentum  [TeV/$c$]")
a1.set_ylabel("ratio to\nthe release curve", fontsize=9)
a1.grid(True, which="both", alpha=0.15)

save_figure(fig, "00_efficiency_products")
plt.show()
```

## Exposure

Two numbers, and they are different. The dataframe's per-segment `exposure_s`
sums to the **full-run** live-time (≈ 469.7 h over 3122 segments). The limit
scans assume the **night-selection** live-time, `config.T_EXPOSURE`
(= 790778 s = 219.66 h, overridable via `LUHDM_T_EXPOSURE`), so that the
notebooks and the cluster scripts share a single value, and the release records
it as `/detector/exposure_s`.

The full-run half of that comparison is the one block here that needs the raw
instrument file; without it the cell prints the released live-time and skips
the rest.

```python
print("NIGHT SELECTION (what the limit scans assume)")
print(f"config.T_EXPOSURE      : {config.T_EXPOSURE:.6g} s")
print(f"                       = {config.T_EXPOSURE / 3600:.2f} h")
print(f"release attribute      : {float(rel.attrs['t_exposure_s']):.6g} s")
assert np.isclose(float(rel.attrs["t_exposure_s"]), config.T_EXPOSURE), \
    "the release was built at a different live-time than config.T_EXPOSURE"

if not HAVE_RAW:
    df = None
    print(f"\nFULL RUN: skipped, {RAW.name} is not on this machine")
else:
    import pandas as pd

    df = pd.read_hdf(RAW)
    exposure_s = df["exposure_s"].sum()
    t0 = pd.to_datetime(df["start_time"]).min()
    t1 = pd.to_datetime(df["end_time"]).max()
    calendar_s = (t1 - t0).total_seconds()
    print("\nFULL RUN (this raw file)")
    print(f"segments               : {len(df)}")
    print(f"exposure (from df)     : {exposure_s:.3e} s")
    print(f"                       = {exposure_s / 3600:.1f} h")
    print(f"                       = {exposure_s / 86400:.2f} days")
    print(f"calendar span          : {t0:%Y-%m-%d} → {t1:%Y-%m-%d} "
          f"({calendar_s / 86400:.1f} days)")
    print(f"duty cycle             : {exposure_s / calendar_s:.1%}")
    print(f"night fraction of it   : {config.T_EXPOSURE / exposure_s:.1%}")
```

## Blips: reconstructed momentum (full run)

For every up-crossing, the blip momentum is taken at its most significant sample
(peak $q$): $p = |\hat{dv}|\,K_n\,c/e$, with $K_n$ the volt→momentum factor
recovered above. The result is one momentum per blip, per mode, in eV/$c$ —
over **every** segment in the raw file, i.e. the full run. Those momenta ship in
the release as `/detector/all_blips_mode{n}`, so the figure below needs no raw
file; where the raw file *is* present, the cell re-derives them from the
transients and checks the two agree.

All of them clear the 100 GeV reconstruction threshold by a wide margin — the
softest blip in the run is 0.47 TeV — so the count that matters for the limit is
how many also fall inside the 1 TeV analysis window, printed alongside.

```python
def blip_momenta_eV(mode):
    """Reconstructed momentum [eV/c] of every q>100 up-crossing in one mode."""
    K = K_VOLT_TO_SI[mode]
    out = []
    q_col = df[f"transient_q_{mode}"]
    dv_col = df[f"transient_dv_hat_{mode}"]
    up_col = df[f"upcrossings_{mode}"]
    for q, dv, ups in zip(q_col, dv_col, up_col):
        if not len(ups):
            continue
        q = np.asarray(q)
        dv = np.asarray(dv)
        for start_i, end_i in ups:
            k = start_i + int(np.argmax(q[start_i:end_i + 1]))  # peak-q sample
            p_si = abs(dv[k]) * K
            out.append(p_si * EV_PER_SI)
    return np.array(sorted(out))


# ALL_MOMENTA came from the release in the setup cell; with the raw file on hand
# it is re-derived here from the transients and the two are compared, which is
# what pins the volt->momentum calibration above to the released numbers.
if HAVE_RAW:
    _redone = {n: blip_momenta_eV(n) for n in MODES}
    for n in MODES:
        assert np.allclose(_redone[n], ALL_MOMENTA[n], rtol=1e-9), \
            f"mode {n}: blips re-derived from the raw file differ from the release"
    print("blip momenta re-derived from the raw transients match the release "
          "to 1e-9 (all three modes)")
else:
    print("blip momenta taken from the release; the raw transients are not "
          "on this machine, so they are not re-derived")

# every reconstructed blip is a candidate; the analysis window keeps a subset
MOMENTA = {n: ALL_MOMENTA[n][ALL_MOMENTA[n] >= Q_WINDOW_TEV * 1e12]
            for n in MODES}
for n in MODES:
    p = ALL_MOMENTA[n]
    below = int((p < Q_RECO_TEV * 1e12).sum())
    print(f"mode {n}: {p.size:3d} blips, "
          f"{below} below {Q_RECO_TEV:g} TeV reconstruction, "
          f"{MOMENTA[n].size:3d} inside the {Q_WINDOW_TEV:g} TeV window, "
          f"p = {p.min() / 1e12:.2f} .. {p.max() / 1e12:.1f} TeV "
          f"(median {np.median(p) / 1e12:.2f} TeV)")
```

## Blip momentum spectrum (full run)

**This figure is the full run, not the night selection.** It is the only place
the blips below the analysis window appear; the release ships them in
`/detector/all_blips_mode{n}`, which carries no night flag. The legend counts are therefore all the full-run
blips (66/99/443), *not* the 8/26/126 night candidates the limits are set on.
Making this panel night-consistent needs the segment-level selection
(start-hour window, the 19–20 Jan exclusion, the lock-in-drive veto) applied to

The shaded band is momentum below the 1 TeV analysis window: reconstructed and

## Per-mode candidate files

`data_mode{n}.txt` holds each mode's night-selected candidates (impulses in eV,
one per line, every one reconstructed above the 0.1 TeV threshold) — the
per-mode analysis inputs the limit scans read via
`scan_grid.py --mode n --data data_mode{n}.txt`. The files are deliberately not
pre-cut to the analysis window: the scan reads the whole list and lays its
momentum grid from `config.Q_THRESH` = 1 TeV upwards, so anything below the
window falls outside the optimum interval's support and contributes nothing to
the limit. Moving the window is then a change of one number, not a
regeneration of the data. The files are produced by the night-selection
pipeline, **not** by this notebook: rerunning it must not overwrite them with
the full-run population, so the cell below only reads them back and checks them.

```python
# These files are NOT written here: they are the night-selection product and
# this notebook only sees the full run (see the note above). Read them back and
# check they are self-consistent instead.
for n in MODES:
    p = CANDIDATES[n]
    assert p.size and np.all(p >= Q_RECO_TEV * 1e12), \
        f"data_mode{n}.txt has an entry below the {Q_RECO_TEV:g} TeV reconstruction threshold"
    in_window = p[p >= Q_WINDOW_TEV * 1e12]
    print(f"data_mode{n}.txt  {p.size:3d} candidates, "
          f"p = {p.min() / 1e12:.2f} .. {p.max() / 1e12:.1f} TeV "
          f"(median {np.median(p) / 1e12:.2f} TeV); "
          f"{in_window.size:3d} inside the {Q_WINDOW_TEV:g} TeV window")

# the ones the limit never sees, named explicitly
_outside = {n: CANDIDATES[n][CANDIDATES[n] < Q_WINDOW_TEV * 1e12] / 1e9
            for n in MODES}
print()
print("below the analysis window [GeV]:",
      {n: np.round(v, 0) for n, v in _outside.items()})

# show data_mode1.txt verbatim as an example
print()
print(Path('data_mode1.txt').read_text()[:600], '...')
```
